```python
ID_COLUMN = "comment_id"
TEXT_COLUMN = "comment_text"

emotion_columns = [
    col for col in df.columns
    if col not in [ID_COLUMN, TEXT_COLUMN]
]

df = pd.read_parquet('./data.parquet')
```

Emotion Score Range

```python
range_df = pd.DataFrame({
    "Minimum": df[emotion_columns].min(),
    "Maximum": df[emotion_columns].max()
})
```

```python
invalid = {}

for col in emotion_columns:
    count = ((df[col] < 0) | (df[col] > 1)).sum()
    if count > 0:
        invalid[col] = count

if len(invalid) == 0:
    print("✓ All emotion scores are within the range [0, 1].")
else:
    print("Columns containing invalid values:")
    for col, count in invalid.items():
        print(f"{col}: {count}")
```

```python
non_numeric = []

for col in emotion_columns:
    if not pd.api.types.is_numeric_dtype(df[col]):
        non_numeric.append(col)

if len(non_numeric) == 0:
    print("✓ All emotion columns are numeric.")
else:
    print("Non-numeric emotion columns:")
    for col in non_numeric:
        print(col)
```
