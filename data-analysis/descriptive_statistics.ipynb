```python
import pandas as pd
```

```python
path = "https://ytcommentstorage.blob.core.windows.net/transformed-data/llm_cleaned.parquet?sv=2026-02-06&ss=bfqt&srt=sco&sp=rwdlacupiytfx&se=2026-07-17T22:04:00Z&st=2026-06-17T13:49:00Z&spr=https&sig=kiL%2BVdjsVrc0PRmMNVAmLuG%2BpYEqTJpC%2BufjbP8dJU4%3D"
df = pd.read_parquet(path)

ID_COLUMN = "comment_id"
TEXT_COLUMN = "comment_text"

emotion_columns = [
    col for col in df.columns
    if col not in [ID_COLUMN, TEXT_COLUMN]
]
```

```python
stats = pd.DataFrame({
    "Mean": df[emotion_columns].mean(),
    "Median": df[emotion_columns].median(),
    "Std Dev": df[emotion_columns].std(),
    "Variance": df[emotion_columns].var(),
    "Minimum": df[emotion_columns].min(),
    "25%": df[emotion_columns].quantile(0.25),
    "50%": df[emotion_columns].quantile(0.50),
    "75%": df[emotion_columns].quantile(0.75),
    "Maximum": df[emotion_columns].max(),
    "Skewness": df[emotion_columns].skew(),
    "Kurtosis": df[emotion_columns].kurt()
})

stats
```

Sorted Mean Score

```python
mean_scores = stats["Mean"].sort_values(ascending=False)
mean_scores
```
