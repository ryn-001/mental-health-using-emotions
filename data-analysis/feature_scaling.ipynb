```python
from sklearn.preprocessing import StandardScaler
import joblib
import pandas as pd
```

```python
ID_COLUMN = "comment_id"
TEXT_COLUMN = "comment_text"

emotion_columns = [
    col for col in df.columns
    if col not in [ID_COLUMN, TEXT_COLUMN]
]

X = df[emotion_columns]

print(X.shape)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=emotion_columns,
    index=df.index
)

X_scaled.head()
```

```python
scaled_summary = pd.DataFrame({
    "Mean": X_scaled.mean(),
    "Std Dev": X_scaled.std()
})

scaled_summary
```

```python
print("Average Mean :", X_scaled.mean().mean())
print("Average Std  :", X_scaled.std().mean())
```

```python
joblib.dump(
    scaler,
    "standard_scaler.pkl"
)
```

```python
scaled_df = pd.concat(
    [
        df[[ID_COLUMN, TEXT_COLUMN]],
        X_scaled
    ],
    axis=1
)

scaled_df.to_parquet(
    "scaled_emotion_dataset.parquet",
    index=False
)

scaled_df.head()
```
